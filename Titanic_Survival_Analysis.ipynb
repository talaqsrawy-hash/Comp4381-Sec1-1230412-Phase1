# Titanic Survival Analysis

## Project Overview

This project explores the Titanic dataset to understand which factors may have affected passengers' chances of survival.

The main goal is to explore the data, answer questions, and find patterns related to passenger survival.

## Project Objectives

- Explore the Titanic dataset.
- Study the relationship between passenger characteristics and survival.
- Identify the factors most related to survival.
- Use charts and statistics to understand patterns in the data.

## Dataset

The dataset contains information about Titanic passengers, such as age, gender, passenger class, ticket fare, and survival status.
